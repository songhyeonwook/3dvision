## 0.2 체커보드 — 인쇄하지 말고 모니터에 띄우기
인쇄는 프린터 · 용지 설정에 따라 크기가 달라짐. 모니터는 평면이 보장되고 칸 크기가 **칸의 픽셀 수 × 픽셀 피치** 로 정확히 정해짐.
1. **24인치 이상 외부 모니터 권장**(노트북 화면은 칸이 12 mm 수준으로 작아 촬영 거리 범위가 좁아짐). 모니터를 **네이티브 해상도**로 (Windows: 설정 → 시스템 → 디스플레이 → 해상도 "권장" / macOS: 시스템 설정 → 디스플레이 → "디스플레이 기본값"). 밝기 최대, 자동 밝기 · 화면 보호기 · 절전 끔
2. 패키지의 **`checkerboard_screen.html`** 을 브라우저로 열고 F11(전체 화면). 패널에 모니터 대각선(inch)을 넣으면 픽셀 피치 = 대각선 × 25.4 / √(W² + H²) 와 **칸 크기(mm)** 가 계산됨. 보드 아래 1000 px 눈금자(100 px 간격 흑백 막대)를 자로 재어 ±1 mm 이내인지 확인. H 키로 패널을 숨기고 촬영
3. 아래 0.6 에 monitor_native_px · monitor_diagonal_inch · pixel_pitch_mm · board_square_px · **board_square_mm** 를 기록 — 코너 검출과 모든 Part 가 이 값을 참조함
4. 촬영 요령: 조명 반사가 없는 각도(실내등 끄기), 셔터 1/50 s 이상으로 화면 깜빡임 줄무늬 방지, 흰 칸이 포화되지 않게(ISO 낮게), 모아레가 보이면 거리를 조금 바꿈, 팔꿈치 고정 · 셀프타이머로 흔들림 방지 (거치형 셀카봉 등 사용 권장)
5. 보드(모니터)는 기울이기 어려우니 **카메라를 움직여** 상대 자세를 만듦 — 세트 B · C 의 "기울기" 는 카메라를 비스듬히 두는 것

## 0.7 코너 검출과 커버리지 맵 (세트 A · B · C)

`scale` 로 전체 해상도 대신 축소해 검출할 수 있지만 **세트 전체에 같은 값**을 써야 하고, 보고서에 기록해야 합니다 (K 가 바뀜 — Part 1.3). Part 3–5는 저장된 `image_size`로 K를 원본 사진 크기에 맞추며, Part 6의 f 비교표도 원본 해상도로 통일합니다. 사진의 화면비·방향·크롭은 바꾸지 마세요.

```python
SCALE = cfg['SCALE']  # hw1_config.json의 scale: 기본 1.0, 느리면 0.5 (모든 세트 동일)
assert meas.get('board_square_mm'), 'measurements.json 의 board_square_mm 를 먼저 채우세요 (0.2 · 0.6)'
dets = {}
for s in ['setA', 'setB', 'setC']:
    d = Bd.detect_set(os.path.join(DATA, s), scale=SCALE, verbose=False, square=meas['board_square_mm'], pattern=tuple(meas['board_inner_corners']))
    Bd.save_corners(IO.rpath(f'corners_{s}.npz'), d)
    dets[s] = d
    print(f'{s}: 검출 {len(d["names"])} 장 / 실패 {len(d["failed"])} 장 {d["failed"]}  image_size={d["image_size"]}')
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))
for ax, s in zip(axes, dets):
    cnt = Bd.coverage_map(dets[s]['imgpoints'], dets[s]['image_size'], ax=ax, title=f'{s} 코너 누적 ({len(dets[s]["names"])}장)')
    print(s, '3×3 구역별 코너 수:\n', cnt)
fig.tight_layout(); fig.savefig(IO.rpath('coverage_ABC.png'), dpi=120)
```
